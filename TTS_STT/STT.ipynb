```python
import whisper 
```

```python
model = whisper.load_model("base") 
```

```python
import os

folder_path = "../TTS_STT/audio"

fichiers = [f for f in os.listdir(folder_path)]

latest = max(fichiers, key=lambda f: os.path.getmtime(os.path.join(folder_path, f)))

print(latest)
```

```python
result = folder_path +"/"+ latest
result
```

```python
result = model.transcribe(result)
print(result["text"])
```
